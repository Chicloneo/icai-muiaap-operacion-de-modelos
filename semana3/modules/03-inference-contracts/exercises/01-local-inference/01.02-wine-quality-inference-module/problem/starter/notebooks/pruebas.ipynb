```python
import csv
import joblib

with open("../../assets/inference_samples.csv", newline="", encoding="utf-8") as file:
    row = next(csv.DictReader(file))

artifact = joblib.load("wine_quality_classifier.joblib")
estimator = artifact["estimator"]
features = artifact["feature_names"]
vector = [[float(row[name]) for name in features]]

print("Modelo usado:", type(estimator).__name__)
print("Orden de entrada:", features)
print("Etiquetas posibles:", estimator.classes_)
print("Predicción:", estimator.predict(vector))
print("Probabilidades:", estimator.predict_proba(vector))
print("Versión del modelo:", artifact["model_version"])
```

```python
FEATURE_NAMES = [
    "fixed_acidity",
    "volatile_acidity",
    "citric_acid",
    "residual_sugar",
    "chlorides",
    "free_sulfur_dioxide",
    "total_sulfur_dioxide",
    "density",
    "ph",
    "sulphates",
    "alcohol",
]

def preprocess_wine_request(wine_request):
   return [getattr(wine_request, name) for name in FEATURE_NAMES]
```

```python
lista_preprocesada = preprocess_wine_request(request)
```
